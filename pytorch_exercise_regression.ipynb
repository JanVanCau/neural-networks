```python
X = df.drop('sales', axis=1)
y = df['sales']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)
```

```python
scaler = StandardScaler()

scaled_X_train = scaler.fit_transform(X_train)
scaled_X_test = scaler.transform(X_test)
```

```python
dataset_train = TensorDataset(torch.tensor(scaled_X_train), torch.tensor(y_train.to_numpy()))
dataset_test = TensorDataset(torch.tensor(scaled_X_test), torch.tensor(y_test.to_numpy()))
```

```python
dataloader_train = DataLoader(dataset_train, batch_size=10, shuffle=True)

counter = 0

for batch_inputs, batch_labels in dataloader_train:
    print('batch_inputs:', batch_inputs)
    print('batch_labels:', batch_labels)
    print('')
    counter += 1
    if counter == 3:
        break
```

```python
model = nn.Sequential(nn.Linear(3, 32), nn.ReLU(), nn.Linear(32, 16), nn.ReLU(), nn.Dropout(0.1), nn.Linear(16,1))
optimizer = optim.Adam(model.parameters(), lr=0.01)
criterion = nn.MSELoss()
```

```python
num_epochs = 1000

for i in range(num_epochs):
    for features, target in dataloader_train:

        features = features.float()
        target = target.float()
        target = target.unsqueeze(1)

        optimizer.zero_grad()
        
        prediction = model(features)    
        loss = criterion(prediction, target)    
        loss.backward()
        optimizer.step()

    if (i+1) % 100 == 0:
        print(f'{round((i+1)/num_epochs*100,2)} % done.')
```

```python
dataloader_test = DataLoader(dataset_test, batch_size=len(dataset_test), shuffle=False)
model.eval()

mse_loss = 0.0

with torch.no_grad():
  for features, target in dataloader_test:
      features = features.float()
      target = target.float()
      target = target.unsqueeze(1)
      
      outputs = model(features)
      loss = criterion(outputs, target)
      mse_loss += loss.item()
      
mse_test = mse_loss/len(dataloader_test)
print(f'RMSE: {round(mse_test**(1/2),2)}')
```
